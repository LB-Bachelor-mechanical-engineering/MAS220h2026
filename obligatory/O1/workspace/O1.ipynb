## Oppgave 1

En hastighetsprofil for en laminær strømning mellom to parallelle plater er gitt ved:

$$\frac{u}{u_{\max}} = 1 - \left(\frac{2y}{h}\right)^2$$

der $h = 0,2\text{ mm}$ er avstanden mellom platene og origo er plassert midt mellom de to platene. Væsken er vann ved $20^\circ\text{C}$. $u_{\max} = 0,2\text{ m/s}$.

* Finn skjærspenningen som virker på den øverste platen.
* Lag en skisse for hvordan skjærspenningen varierer over $h$ (avstanden mellom platene).

```python
# Oppgave 1 - Skjærspenning i laminær strømning

import numpy as np
import matplotlib.pyplot as plt

# --- Gitte verdier ---
h = 0.2e-3          # Avstand mellom plater [m] (0.2 mm)
u_max = 0.2         # Maksimal hastighet [m/s]
mu = 1.002e-3       # Dynamisk viskositet for vann ved 20°C [Pa·s]

# --- Teoretisk utledning ---
# Hastighetsprofil: u(y) = u_max * (1 - (2*y / h)^2)
# Hastighetsderivert: du/dy = u_max * (-8 * y / h^2)
# Skjærspenning: tau(y) = mu * (du/dy) = -8 * mu * u_max * y / h^2

# 1. Skjærspenning på øverste plate (y = +h/2)
y_top = h / 2
du_dy_top = -8 * u_max * y_top / (h**2)
tau_top = mu * du_dy_top

# Skjærspenningen som virker FRA væsken PÅ den øverste platen (reaksjonskraft)
tau_on_top_plate = abs(tau_top)

print(f"Hastighetsderivert ved øvre plate (du/dy): {du_dy_top:.2f} s^-1")
print(f"Skjærspenning i væsken ved øvre plate: {tau_top:.4f} Pa")
print(f"Skjærspenning som virker på øverste plate: {tau_on_top_plate:.4f} Pa")

# --- Plotting av hastighetsprofil og skjærspenning over h ---
y_vals = np.linspace(-h/2, h/2, 200)  # y-posisjoner fra nederste til øverste plate
u_vals = u_max * (1 - (2 * y_vals / h)**2)
tau_vals = mu * (-8 * u_max * y_vals / (h**2))

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1: Hastighetsprofil u(y)
ax[0].plot(u_vals, y_vals * 1e3, color='blue', linewidth=2)
ax[0].axhline(y=h/2*1e3, color='black', linestyle='--', label='Øvre plate')
ax[0].axhline(y=-h/2*1e3, color='black', linestyle='--', label='Nedre plate')
ax[0].axhline(y=0, color='gray', linestyle=':', alpha=0.7)
ax[0].set_xlabel('Hastighet u [m/s]')
ax[0].set_ylabel('Posisjon y [mm]')
ax[0].set_title('Hastighetsprofil u(y)')
ax[0].grid(True)
ax[0].legend()

# Plot 2: Skjærspenningsprofil tau(y)
ax[1].plot(tau_vals, y_vals * 1e3, color='red', linewidth=2)
ax[1].axhline(y=h/2*1e3, color='black', linestyle='--')
ax[1].axhline(y=-h/2*1e3, color='black', linestyle='--')
ax[1].axvline(x=0, color='gray', linestyle=':', alpha=0.7)
ax[1].set_xlabel('Skjærspenning $\\tau$ [Pa]')
ax[1].set_ylabel('Posisjon y [mm]')
ax[1].set_title('Skjærspenning $\\tau(y)$ over plateavstanden h')
ax[1].grid(True)

plt.tight_layout()
plt.show()
```
