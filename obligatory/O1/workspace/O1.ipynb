## Oppgave 1

En hastighetsprofil for en laminær strømning mellom to parallelle plater er gitt ved:

$$\frac{u}{u_{\max}} = 1 - \left(\frac{2y}{h}\right)^2$$

der $h = 0,2\text{ mm}$ er avstanden mellom platene og origo er plassert midt mellom de to platene. Væsken er vann ved $20^\circ\text{C}$. $u_{\max} = 0,2\text{ m/s}$.

* Finn skjærspenningen som virker på den øverste platen.
* Lag en skisse for hvordan skjærspenningen varierer over $h$ (avstanden mellom platene).

```python
# Oppgave 1 - Skjærspenning i laminær strømning

import numpy as np
import matplotlib.pyplot as plt

# --- Gitte verdier ---
h = 0.2e-3          # Avstand mellom plater [m] (0.2 mm)
u_max = 0.2         # Maksimal hastighet [m/s]
mu = 1.002e-3       # Dynamisk viskositet for vann ved 20°C [Pa·s]

# --- Teoretisk utledning ---
# Hastighetsprofil: u(y) = u_max * (1 - (2*y / h)^2)
# Hastighetsderivert: du/dy = u_max * (-8 * y / h^2)
# Skjærspenning: tau(y) = mu * (du/dy) = -8 * mu * u_max * y / h^2

# 1. Skjærspenning på øverste plate (y = +h/2)
y_top = h / 2
du_dy_top = -8 * u_max * y_top / (h**2)
tau_top = mu * du_dy_top

# Skjærspenningen som virker FRA væsken PÅ den øverste platen (reaksjonskraft)
tau_on_top_plate = abs(tau_top)

print(f"Hastighetsderivert ved øvre plate (du/dy): {du_dy_top:.2f} s^-1")
print(f"Skjærspenning i væsken ved øvre plate: {tau_top:.4f} Pa")
print(f"Skjærspenning som virker på øverste plate: {tau_on_top_plate:.4f} Pa")

# --- Plotting av hastighetsprofil og skjærspenning over h ---
y_vals = np.linspace(-h/2, h/2, 200)  # y-posisjoner fra nederste til øverste plate
u_vals = u_max * (1 - (2 * y_vals / h)**2)
tau_vals = mu * (-8 * u_max * y_vals / (h**2))

fig, ax = plt.subplots(1, 2, figsize=(12, 5))

# Plot 1: Hastighetsprofil u(y)
ax[0].plot(u_vals, y_vals * 1e3, color='blue', linewidth=2)
ax[0].axhline(y=h/2*1e3, color='black', linestyle='--', label='Øvre plate')
ax[0].axhline(y=-h/2*1e3, color='black', linestyle='--', label='Nedre plate')
ax[0].axhline(y=0, color='gray', linestyle=':', alpha=0.7)
ax[0].set_xlabel('Hastighet u [m/s]')
ax[0].set_ylabel('Posisjon y [mm]')
ax[0].set_title('Hastighetsprofil u(y)')
ax[0].grid(True)
ax[0].legend()

# Plot 2: Skjærspenningsprofil tau(y)
ax[1].plot(tau_vals, y_vals * 1e3, color='red', linewidth=2)
ax[1].axhline(y=h/2*1e3, color='black', linestyle='--')
ax[1].axhline(y=-h/2*1e3, color='black', linestyle='--')
ax[1].axvline(x=0, color='gray', linestyle=':', alpha=0.7)
ax[1].set_xlabel('Skjærspenning $\\tau$ [Pa]')
ax[1].set_ylabel('Posisjon y [mm]')
ax[1].set_title('Skjærspenning $\\tau(y)$ over plateavstanden h')
ax[1].grid(True)

plt.tight_layout()
plt.show()
```

## Oppgave 2 - Trykk

En mann som veier $90\text{ kg}$, har fotavtrykk areal på en fot $450\text{ cm}^2$. Finn trykket mannen utøver på bakken:

### a)
Hvis han står på en fot

```python
# Oppgave 2a - Trykk når mannen står på én fot

# --- Gitte verdier ---
m = 90.0           # Masse [kg]
g = 9.81           # Tyngdeakselerasjon [m/s^2]
A_one_foot = 450e-4 # Areal for én fot [m^2] (450 cm^2 = 450 * 10^-4 m^2)

# --- Beregning ---
# Kraft utøvd av mannen på bakken (tyngdekraften)
F = m * g          # Kraft [N]

# Trykk når han står på én fot
P_a = F / A_one_foot # Trykk [Pa]
P_a_kPa = P_a / 1000 # Trykk [kPa]

print(f"Normalkraft (vekt): {F:.2f} N")
print(f"Trykk på bakken (én fot): {P_a:.2f} Pa ({P_a_kPa:.2f} kPa)")
```

### b)
Hvis han står på begge føtter.


```python
# Oppgave 2b - Trykk når mannen står på begge føtter

# --- Gitte verdier ---
m = 90.0           # Masse [kg]
g = 9.81           # Tyngdeakselerasjon [m/s^2]
A_one_foot = 450e-4 # Areal for én fot [m^2]

# --- Beregning ---
F = m * g          # Kraft [N]

# Totalt areal for begge føtter
A_both_feet = 2 * A_one_foot # Areal [m^2]

# Trykk når han står på begge føtter
P_b = F / A_both_feet # Trykk [Pa]
P_b_kPa = P_b / 1000 # Trykk [kPa]

print(f"Totalt areal (begge føtter): {A_both_feet:.4f} m^2")
print(f"Trykk på bakken (begge føtter): {P_b:.2f} Pa ({P_b_kPa:.2f} kPa)")
```

## Oppgave 3 - Manometer med flere fluider

Trykket til naturgassen blir målt med manometer vist i figuren nedenfor. Atmosfæretrykket er $98\text{ kPa}$. Finn trykket til naturgassen. Kvikksølv tetthet er $13\,600\text{ kg/m}^3$, tettheten til oljen er $690\text{ kg/m}^3$ og høydene er følgende: $h_{\text{Hg}} = 0,15\text{ m}$, $h_{\text{oil}} = 0,19\text{ m}$ og $h_w = 0,6\text{ m}$.

```python
# Oppgave 3 - Manometer med flere fluider

# --- Gitte verdier ---
P_atm = 98e3        # Atmosfæretrykk [Pa] (98 kPa)
g = 9.81            # Tyngdeakselerasjon [m/s^2]

# Tettheter [kg/m^3]
rho_Hg = 13600.0    # Kvikksølv
rho_oil = 690.0     # Olje
rho_water = 1000.0  # Vann

# Høydedifferanser [m]
h_Hg = 0.15         # Kvikksølvsøyle
h_oil = 0.19        # Oljesøyle
h_w = 0.60          # Vannsøyle

# --- Beregning ---
# Vi følger trykket fra gassen gjennom manometeret til den åpne enden:
# P_gas + rho_Hg * g * h_Hg - rho_oil * g * h_oil + rho_water * g * h_w = P_atm
# Dermed blir absolutt trykk for naturgassen:
P_gas_abs = P_atm - rho_Hg * g * h_Hg + rho_oil * g * h_oil - rho_water * g * h_w

# Overtrykk (gauge pressure)
P_gas_gauge = P_gas_abs - P_atm

print(f"Absolutt trykk for naturgassen: {P_gas_abs:.2f} Pa ({P_gas_abs/1000:.2f} kPa)")
print(f"Overtrykk (gauge) for naturgassen: {P_gas_gauge:.2f} Pa ({P_gas_gauge/1000:.2f} kPa)")
```

## Oppgave 4 - Hydrostatiske krefter på plane flater

En port i en dam har en horisontal og en vertikal del, som vist i figuren. Porten er $3\text{ m}$ bred og er hengslet ved O, massen er neglisjerbar. Porten er forhindret fra å rotere mot urviseren. Væsken er ferskvann med tetthet lik $1000\text{ kg/m}^3$. Vannfyllingen, $H$, er lik $2\text{ m}$.

Annet arealmoment om en akse gjennom flatesenteret er for et rektangulært tverrsnitt gitt ved:

$$I_{xx,C} = \frac{1}{12}bh^3$$

### a)
Bestem hydrostatisk kraft på den horisontale delen av porten. Hva er angrepspunktet (virkelinjen) til denne kraften?

*NB! Lengden/dybden til porten innover i papirplanet kan antas å være $1\text{ m}$ for enkelthets skyld!*

```python
# Oppgave 4a - Hydrostatisk kraft på den horisontale delen av porten

# --- Gitte verdier ---
rho = 1000.0      # Tetthet for ferskvann [kg/m^3]
g = 9.81          # Tyngdeakselerasjon [m/s^2]
H = 2.0           # Vannhøyde over den horisontale platen [m]
L_horiz = 3.0     # Lengde på den horisontale delen [m]
b = 1.0           # Dybde innover i papirplanet [m]

# --- Beregning ---
# Areal av den horisontale porsjonen som påvirkes av vannsøylen
A_horiz = L_horiz * b

# Jevnt trykk over den horisontale platen
P_horiz = rho * g * H

# Hydrostatisk kraft (reaktionskraft/vekt av vannet over platen)
F_horiz = P_horiz * A_horiz

# Angrepspunkt (siden trykket er jevnt fordelt, virker kraften i midten)
x_cp = L_horiz / 2.0

print(f"Hydrostatisk kraft på horisontal del: {F_horiz:.2f} N ({F_horiz/1000:.2f} kN)")
print(f"Angrepspunkt (avstand fra hengsel O langs platen): {x_cp:.2f} m")
```

### b)
Bestem hydrostatisk kraft på den vertikale delen av porten. Finn angrepspunktet (virkelinjen) til denne kraften.

```python
# Oppgave 4b - Hydrostatisk kraft på den vertikale delen av porten

# --- Gitte verdier ---
rho = 1000.0      # Tetthet [kg/m^3]
g = 9.81          # [m/s^2]
H = 2.0           # Vannhøyde [m]
b = 1.0           # Dybde inn i papirplanet [m]

# --- Beregning ---
# Våt areal av vertikal plate
A_vert = H * b

# Trykk ved midtpunktet (flatesenter, h_c)
h_c = H / 2.0
P_c = rho * g * h_c

# Hydrostatisk totalkraft
F_vert = P_c * A_vert

# Angrepspunkt (trykksenter, y_cp) målt vertikalt ned fra vannflaten:
# y_cp = h_c + I_xC / (h_c * A) = H/2 + (1/12 * b * H^3) / ((H/2) * (b * H)) = (2/3) * H
y_cp_from_surface = (2 / 3) * H

# Avstand fra hengsel O og opp til angrepspunktet
y_cp_from_O = H - y_cp_from_surface

print(f"Hydrostatisk kraft på vertikal del: {F_vert:.2f} N ({F_vert/1000:.2f} kN)")
print(f"Angrepspunkt fra vannflaten og ned: {y_cp_from_surface:.2f} m")
print(f"Angrepspunkt fra hengsel O og opp: {y_cp_from_O:.2f} m")
```

### c)
Ved hvilken verdi av $H$ vil porten begynne å rotere?

```python
# Oppgave 4c - Kritiske vannhøyde H der porten begynne å rotere

# --- Utledning ---
# Porten er forhindret fra å rotere MOT urviseren. Dvs. den roterer NÅR momentet
# om hengsel O fra kreftene blir med urviseren (M_O > 0).
#
# Moment fra vertikal kraft (roterer MOT urviseren, -):
# F_vert = rho * g * (H^2 / 2) * b
# Arm fra O = H / 3
# M_vert = F_vert * (H / 3) = rho * g * b * H^3 / 6
#
# Moment fra horisontal kraft (roterer MED urviseren, +):
# F_horiz = rho * g * H * L * b
# Arm fra O = L / 2
# M_horiz = F_horiz * (L / 2) = rho * g * b * H * L^2 / 2
#
# Ved vippepunktet gjelder: M_vert = M_horiz
# (rho * g * b * H^3) / 6 = (rho * g * b * H * L^2) / 2
# H^2 / 3 = L^2 / 2  =>  H = L * sqrt(3 / 2)

import numpy as np

L_horiz = 3.0  # [m]

H_crit = L_horiz * np.sqrt(3 / 2)

print(f"Kritisk vannhøyde H der porten begynner å rotere: {H_crit:.3f} m")
```

### d)
Når $H$ er $6\text{ m}$, hvilket moment må påsettes i O for at porten skal være i likevekt? Størrelsen og retningen til momentet skal bestemmes.

```python
# Oppgave 4d - Nødvendig moment i O for likevekt når H = 6 m

# --- Gitte verdier ---
rho = 1000.0      # [kg/m^3]
g = 9.81          # [m/s^2]
H = 6.0           # Ny vannhøyde [m]
L_horiz = 3.0     # [m]
b = 1.0           # [m]

# --- Beregning av momenter om O ---
# 1. Moment fra den vertikale væskekraften (roterer mot urviseren)
F_vert = rho * g * (H**2 / 2) * b
arm_vert = H / 3.0
M_vert = F_vert * arm_vert  # [Nm]

# 2. Moment fra den horisontale væskekraften (roterer med urviseren)
F_horiz = rho * g * H * L_horiz * b
arm_horiz = L_horiz / 2.0
M_horiz = F_horiz * arm_horiz  # [Nm]

# Netto hydrostatisk moment om O (positiv retning definert MED urviseren)
M_netto = M_horiz - M_vert

# Påsatt ytre moment M_O må motvirke netto hydrostatisk moment
M_O = -M_netto

retning = "med urviseren" if M_O > 0 else "mot urviseren"

print(f"Moment fra vertikal flate:   {M_vert:.2f} Nm")
print(f"Moment fra horisontal flate: {M_horiz:.2f} Nm")
print(f"Nødvendig ytre moment M_O i O: {abs(M_O):.2f} Nm ({abs(M_O)/1000:.2f} kNm), roterende {retning}")
```

## Oppgave 5 - Hydrostatiske krefter på krumme flater

En $70\text{ meter}$ lang demning har fasongen til en kvart sirkel med radius på $7\text{ m}$ ($R = 7\text{ m}$).

### a)
Finn den horisontale kraften ($F_H$)

```python
# Oppgave 5a - Horisontal kraft på krum flate

# --- Gitte verdier ---
rho = 1000.0      # Tetthet for vann [kg/m^3]
g = 9.81          # Tyngdeakselerasjon [m/s^2]
R = 7.0           # Radius på kvart sirkel [m]
L = 70.0          # Lengde på demningen [m]

# --- Beregning ---
# Projisert areal i det vertikale planet
A_proj_vert = R * L

# Dypt til flatesenter for det projiserte arealet (h_c)
h_c = R / 2.0

# Den horisontale kraften er lik kraften på det projiserte vertikale arealet
F_H = rho * g * h_c * A_proj_vert

print(f"Projisert vertikalt areal: {A_proj_vert:.2f} m^2")
print(f"Horisontal kraft F_H: {F_H:.2f} N ({F_H/1e6:.3f} MN)")
```

### b)
Finn den vertikale kraften ($W$)

```python
# Oppgave 5b - Vertikal kraft på krum flate

import numpy as np

# --- Gitte verdier ---
rho = 1000.0      # Tetthet for vann [kg/m^3]
g = 9.81          # [m/s^2]
R = 7.0           # Radius [m]
L = 70.0          # Lengde [m]

# --- Beregning ---
# Den vertikale kraften W tilsvarer tyngden av vannvolumet som ligger over den krumme flaten.
# Tverrsnittet er en kvart sirkel med areal: A_kvartsirkel = (pi * R^2) / 4
A_cross = (np.pi * R**2) / 4.0

# Totalt vannvolum over flaten
V = A_cross * L

# Vertikal kraft (tyngde av vannvolum)
F_V = rho * g * V

print(f"Vannvolum over flaten: {V:.2f} m^3")
print(f"Vertikal kraft F_V (W): {F_V:.2f} N ({F_V/1e6:.3f} MN)")
```

### c)
Hva blir totalkraften.

```python
# Oppgave 5c - Totalkraft på krum flate

import numpy as np

# --- Gitte verdier ---
rho = 1000.0      # [kg/m^3]
g = 9.81          # [m/s^2]
R = 7.0           # [m]
L = 70.0          # [m]

# --- Komponenter ---
F_H = rho * g * (R / 2.0) * (R * L)
F_V = rho * g * ((np.pi * R**2) / 4.0) * L

# Resultante / Totalkraft (pythagoras)
F_total = np.sqrt(F_H**2 + F_V**2)

# Vinkel theta målt i forhold til horisontalaksen
theta_rad = np.arctan(F_V / F_H)
theta_deg = np.degrees(theta_rad)

print(f"Totalkraft F_resultat: {F_total:.2f} N ({F_total/1e6:.3f} MN)")
print(f"Vinkel mot horisontalplanet: {theta_deg:.2f}°")
```

## Oppgave 6 - Oppdrift

En $200\text{ kg}$ granittstein (tetthet til steinen er $2700\text{ kg/m}^3$) mistes i en innsjø (vann tetthet $1000\text{ kg/m}^3$). En mann dykker ned for å løfte den opp.

### a)
Tegne fri legemediagram

```python
# Oppgave 6a - Fri legemediagram (vektorkrefter)

# Et fri legemediagram for steinen under vann inneholder følgende tre vertikale krefter:
# 1. Tyngdekraften (W = m * g) virker NEDOVER.
# 2. Oppdrift (F_B = rho_vann * V * g) virker OPPOVER.
# 3. Løftekaft fra dykkeren (F_løft) virker OPPOVER.

# Siden systemet er i statisk likevekt ved heving uten akselerasjon, gjelder:
# sum(F_y) = 0  =>  F_løft + F_B - W = 0

import matplotlib.pyplot as plt

# Visualisering av fri legemediagram
fig, ax = plt.subplots(figsize=(5, 6))

# Tegn steinen
stone = plt.Circle((0, 0), 0.3, color='gray', ec='black', lw=2)
ax.add_patch(stone)
ax.text(0, 0, 'Granittstein\n(m = 200 kg)', ha='center', va='center', fontweight='bold')

# Vektorer (krefter)
# Tyngdekraft (ned)
ax.annotate('', xy=(0, -0.9), xytext=(0, -0.3),
            arrowprops=dict(facecolor='red', edgecolor='red', shrink=0, width=2, headwidth=8))
ax.text(0.05, -0.65, '$W = m \cdot g$ (Tyngde)', color='red', fontsize=11)

# Oppdrift (opp)
ax.annotate('', xy=(-0.1, 0.8), xytext=(-0.1, 0.3),
            arrowprops=dict(facecolor='blue', edgecolor='blue', shrink=0, width=2, headwidth=8))
ax.text(-0.55, 0.55, '$F_B$ (Oppdrift)', color='blue', fontsize=11)

# Løftekraft (opp)
ax.annotate('', xy=(0.1, 0.6), xytext=(0.1, 0.3),
            arrowprops=dict(facecolor='green', edgecolor='green', shrink=0, width=2, headwidth=8))
ax.text(0.15, 0.45, '$F_{løft}$ (Fra dykker)', color='green', fontsize=11)

ax.set_xlim(-1, 1)
ax.set_ylim(-1.2, 1.2)
ax.set_aspect('equal')
ax.axis('off')
plt.title('Fri legemediagram (FBD)')
plt.show()
```

### b)
Finn kraften som trengs for å løfte den opp? Kan en dykker klare det?

```python
# Oppgave 6b - Løftekraft og vurdering

# --- Gitte verdier ---
m_stein = 200.0       # Masse til steinen [kg]
rho_stein = 2700.0    # Tetthet for granitt [kg/m^3]
rho_vann = 1000.0     # Tetthet for vann [kg/m^3]
g = 9.81              # Tyngdeakselerasjon [m/s^2]

# --- Beregning ---
# 1. Tyngdekraft
W = m_stein * g

# 2. Volum av steinen (V = m / rho)
V_stein = m_stein / rho_stein

# 3. Oppdrift i vann (Arkimedes' lov)
F_B = rho_vann * g * V_stein

# 4. Nødvendig løftekraft for likevekt (F_løft = W - F_B)
F_loft = W - F_B

# Vekt som denne kraften tilsvarer i "effektiv masse" på land
m_effektiv = F_loft / g

print(f"Tyngdekraft i luft (W):    {W:.2f} N")
print(f"Oppdrift fra vannet (F_B):  {F_B:.2f} N")
print(f"Nødvendig løftekraft:      {F_loft:.2f} N ({F_loft/1000:.2f} kN)")
print(f"Effektiv masse under vann: {m_effektiv:.1f} kg")

# --- Vurdering ---
# Normalt kan et menneske løfte ca. 20-50 kg under vann uten spesialutstyr.
grense_dykker_kg = 50.0

print("\n--- Vurdering av om dykkeren klarer det ---")
if m_effektiv <= grense_dykker_kg:
    print(f"Ja, dykkeren klarer det alene da belastningen tilsvarer {m_effektiv:.1f} kg.")
else:
    print(f"Nei, {m_effektiv:.1f} kg er for tungt for en dykker uten løfteballong/heveutstyr.")
```
